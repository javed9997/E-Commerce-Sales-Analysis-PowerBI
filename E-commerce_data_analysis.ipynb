```python
# Step 1:Import Required Libraries

import pandas as pd
import numpy as np
```

```python
# Step 2: Load Dataset

df = pd.read_csv('PowerBI_Ecommerce_Orders.csv')
```

```python
# Step 3: Explore the Dataset
# View First 5 Records

df.head()
```

```python
# Step 4: Dataset Information
df.info()
```

```python
# Dataset Shape

df.shape
```

```python
# Step 4: check Missing Values

df.isnull().sum()
```

```python
# Step 5: Check Duplicate Records

df.duplicated().sum()
```

```python
# Remove Duplicate Records

df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
# Step 6: Replace Invalid Values
# Replace common invalid values with NaN

df.replace(['N/A','NULL',''],np.nan,inplace=True)
```

```python
df
```

```python
# Step 7: Remove Extra Spaces

df = df.apply(lambda x: x.str.strip() if x.dtype =='object' else x)
```

```python
# Step 8: Standardize Text Format
# Convert text into Proper Case.

df['Customer_Name'] = df['Customer_Name'].str.title()

df['City'] = df['City'].str.title()

df['State'] = df['State'].str.title()
```

```python
# Step 9: Validate Email records
# Keep only valid Email records.

df = df[df['Email'].str.contains('@',na=False)]
```

```python
# Step 10 Convert Date Columns

df['Order_Date'] = pd.to_datetime(
    df['Order_Date'],
    format='%d-%m-%Y',
    errors='coerce'
)

df['Delivery_Date'] = pd.to_datetime(
    df['Delivery_Date'],
    format='%d-%m-%Y',
    errors='coerce'
)
```

```python
# Step 11: Convert Numeric Columns

cols = ['Qty','Unit_Price','Discount']

for c in cols:
    df[c] = pd.to_numeric(df[c],errors='coerce')
```

```python
# Step 12: Remove Invalid Quantity
# Remove records having negative or zero quantity.

df = df[df['Qty']>0]
```

```python
# Step 13: Fill Missing Values

df['Discount'] = df['Discount'].fillna(0)

df['Phone'] = df['Phone'].fillna('Unknown')

df['Delivery_Date'] = df['Delivery_Date'].fillna(df['Order_Date'])
```

# Featuer Engineering

```python
# Total Sales

df['Sales']=df['Qty']*df['Unit_Price']
```

```python
# Net Amount

df['Net_Amount']=df['Sales']-(df['Sales']*df['Discount']/100)
```

```python
df
```

```python
# Profit(Example)
df['Profit']=df['Net_Amount']*0.20
```

```python
# Order Month

df['Month']=df['Order_Date'].dt.month_name()
```

```python
# Order Year

df["Year"]=df['Order_Date'].dt.year
```

```python
# Weekday

df["Weekday"]=df['Order_Date'].dt.day_name()
```

```python
# Step 16: Export Clean Data
df.to_csv('Clean_Ecommerce.csv',index=False)
```
